```python
from pydicom import dcmread
from matplotlib import pyplot as plt
import numpy as np
```

```python
x_start = 139
x_end = 415
y_start = 179
y_end = 287
sizex = x_end-x_start
sizey = y_end-y_start
```

```python
def top(index):
    filename = 'AlokKumar_2023821602_CD/I'+str(index)
    ds = pydicom.dcmread(filename)
    arr = ds.pixel_array
    x_start = 139
    x_end = 415
    y_start = 179
    y_end = 287
    for i in range (x_start-2,x_end+2):
        for j in range(y_start-2,y_end+2):
            arr[i][j]=2000
    ds.PixelData = arr.tobytes()
    ds.save_as("mod/I"+str(index))
```

```python
def others(index):
    filename = 'AlokKumar_2023821602_CD/I'+str(index)
    ds = pydicom.dcmread(filename)
    arr = ds.pixel_array
    for i in range(-2,2):
        for j in range(-2,2):
            arr[x_start+i][y_start+j]=2000
            arr[x_start+sizex+i][y_start+j]=2000
            arr[x_start+i][y_start+sizey+j]=2000
            arr[x_start+sizex+i][y_start+sizey+j]=2000
    ds.PixelData = arr.tobytes()
    ds.save_as("mod/I"+str(index))
```

```python
for i in range(1110,3720,10):
	others(i)
top(1110)
top(3720)
```
