```python
import pydicom
import matplotlib.pyplot as plt
```

```python
x_start = 139
x_end = 415
y_start = 179
y_end = 287
z_start=1100    #start slice
z_end=3750  #end slice
```

```python
def transform_1(index):
    filename = 'AlokKumar_2023821602_CD/I'+str(index)
    ds = pydicom.dcmread(filename)
    arr = ds.pixel_array
    for i in range(0,512):
        for j in range(0,512):
            arr[i][j]=0
    ds.PixelData = arr.tobytes()
    ds.save_as("mod/I"+str(index))
```

```python
def transform_2(index):
    filename = 'AlokKumar_2023821602_CD/I'+str(index)
    ds = pydicom.dcmread(filename)
    arr = ds.pixel_array
    for i in range(0, 512):
        for j in range(0, 512):
            if i>=x_start and i<=x_end and j>=y_start and j<=y_end:
                arr[i][j]=arr[i][j]
            else:
                arr[i][j]=0
    ds.PixelData = arr.tobytes()
    ds.save_as("mod/I"+str(index))
```

```python
for i in range(10,4490,10):
    if i>=z_start and i<=z_end:
        transform_2(i)
    else:
        transform_1(i)
```
